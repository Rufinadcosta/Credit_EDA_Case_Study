# I am using 3 types of plots for analysis. 
- Pie Plot: Where the columns have values interms of percentage
- Countplot1 : Where the data is plotted interms of count of different category
- Countplot2: Where the data is first divided into two subset, then again the subsets are divided into different category. Then again these categories are plotted iterms of percentage

```python
def plotting(train, train0, train1, column):
    
    train = train
    train_0 = train0
    train_1 = train1
    col = column
    
    fig = plt.figure(figsize=(13,10))
    
    axis1 = plt.subplot(221)
    train[col].value_counts().plot.pie(autopct = "%1.0f%%", ax=axis1)
    plt.title('Plotting data for the column: '+ column)
    
    axis2 = plt.subplot(222)
    sns.countplot(x= column, hue = 'TARGET', data = train, ax = axis2)
    plt.xticks(rotation=90)
    plt.title('Plotting data for target in terms of total count')


    axis3 = plt.subplot(223)
    df = pd.DataFrame()
    df['0']= ((train_0[col].value_counts())/len(train_0))
    df['1']= ((train_1[col].value_counts())/len(train_1))
    df.plot.bar(ax=axis3)
    plt.title('Plotting data for target in terms of percentage')


    plt.tight_layout() 
    plt.show()
```

```python
train_categorical_data = application.select_dtypes(include=['object']).columns
train_categorical_data
```

```python
for column in train_categorical_data:
    print("Plotting ", column)
    plotting(application, train_0, train_1, column)
    print('----------------------------------------------------------------------------------------------')
```
